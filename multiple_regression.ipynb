```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# ข้อมูลตัวอย่าง (สมมติ)
# ประสบการณ์(ปี), วุฒิการศึกษา(1=ป.ตรี, 2=โท, 3=เอก), ทักษะภาษาอังกฤษ(0-10), เงินเดือน
data = {
    "experience": [1, 3, 5, 7, 9, 2, 4, 6],
    "education":  [1, 1, 2, 2, 3, 1, 2, 3],
    "english":    [3, 5, 6, 7, 8, 4, 6, 9],
    "salary":     [18000, 22000, 30000, 35000, 50000, 20000, 32000, 55000]
}

df = pd.DataFrame(data)

# แยกตัวแปรอิสระ (X) และตัวแปรตาม (y)
X = df[["experience", "education", "english"]]
y = df["salary"]

# สร้างโมเดล Multiple Regression
model = LinearRegression()
model.fit(X, y)

# แสดงผล
print("ค่าสัมประสิทธิ์ (coefficients):", model.coef_)
print("จุดตัดแกน (intercept):", model.intercept_)

# ทดลองทำนายเงินเดือนของคนใหม่
# สมมติ: ประสบการณ์ 5 ปี, วุฒิ ป.โท (2), ทักษะภาษาอังกฤษ=7
new_person = [[5, 2, 7]]
predicted_salary = model.predict(new_person)
print("เงินเดือนที่คาดการณ์:", predicted_salary[0])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# ข้อมูลตัวอย่าง (สมมติ)
data = {
    "experience": [1, 3, 5, 7, 9, 2, 4, 6],
    "education":  [1, 1, 2, 2, 3, 1, 2, 3],
    "english":    [3, 5, 6, 7, 8, 4, 6, 9],
    "salary":     [18000, 22000, 30000, 35000, 50000, 20000, 32000, 55000]
}

df = pd.DataFrame(data)

# แยกตัวแปรอิสระ (X) และตัวแปรตาม (y)
X = df[["experience", "education", "english"]]
y = df["salary"]

# สร้างโมเดล Multiple Regression
model = LinearRegression()
model.fit(X, y)

# ทำนายค่าที่โมเดลคาดการณ์ได้
y_pred = model.predict(X)

# สร้างกราฟ scatter + line
plt.figure(figsize=(8,6))

plt.scatter(y, y_pred, color="blue", label="Actual vs Predicted")
plt.plot([y.min(), y.max()], [y.min(), y.max()], color="red", linestyle="--", label="Perfect Prediction")

plt.xlabel("Actual Salary")
plt.ylabel("Predicted Salary")
plt.title("Multiple Regression: Actual vs Predicted Salary")
plt.legend()
plt.grid(True)
plt.show()
```
